Checks the environment end to end: build a Bell pair, run it on the ideal and noisy simulators, compare.

```python
from qiskit import QuantumCircuit
from qiskit.quantum_info import SparsePauliOp
from qiskit.visualization import plot_histogram

from qhack import expectation_values, run_counts

bell = QuantumCircuit(2)
bell.h(0)
bell.cx(0, 1)
bell.draw("mpl")
```

```python
measured = bell.measure_all(inplace=False)

ideal = run_counts(measured, "aer", shots=2000)
noisy = run_counts(measured, "noisy", shots=2000)
plot_histogram([ideal, noisy], legend=["ideal", "noisy (device noise model)"])
```

```python
observables = [SparsePauliOp("ZZ"), SparsePauliOp("XX")]
for name in ("aer", "noisy"):
    zz, xx = expectation_values(bell, observables, name)
    print(f"{name:>6}:  <ZZ> = {zz:+.3f}   <XX> = {xx:+.3f}")
```
